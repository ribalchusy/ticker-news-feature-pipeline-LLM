```python
from pathlib import Path
import pandas as pd
import os

DATA_DIR = Path("../data/external/kaggle/russian_financial_news")

files = sorted([file for file in DATA_DIR.rglob("*") if file.is_file()])

print("Files count:", len(files))

for file in files:
    size_mb = file.stat().st_size / 1024 / 1024
    print(f"{file.relative_to(DATA_DIR)} | {size_mb:.2f} MB")
```

```python
csv_files = sorted(DATA_DIR.rglob("*.csv"))
json_files = sorted(DATA_DIR.rglob("*.json"))
parquet_files = sorted(DATA_DIR.rglob("*.parquet"))
xlsx_files = sorted(DATA_DIR.rglob("*.xlsx"))

print("CSV:", len(csv_files))
print("JSON:", len(json_files))
print("PARQUET:", len(parquet_files))
print("XLSX:", len(xlsx_files))

print("\nCSV files:")
for file in csv_files:
    print(file.relative_to(DATA_DIR))
```

```python
for file in csv_files:
    print("\n" + "=" * 120)
    print(file.relative_to(DATA_DIR))
    
    try:
        df_sample = pd.read_csv(file, nrows=5)
        print("shape sample:", df_sample.shape)
        print("columns:", df_sample.columns.tolist())
        display(df_sample.head())
    except Exception as e:
        print("ERROR with default encoding:", e)
        
        for encoding in ["utf-8", "utf-8-sig", "cp1251"]:
            try:
                df_sample = pd.read_csv(file, nrows=5, encoding=encoding)
                print("worked encoding:", encoding)
                print("shape sample:", df_sample.shape)
                print("columns:", df_sample.columns.tolist())
                display(df_sample.head())
                break
            except Exception:
                pass
```

```python
from pathlib import Path
import pandas as pd

events_path = Path("../data/interim/news/news_events_precomputed_llm_2024.parquet")

events = pd.read_parquet(events_path)

print(events.shape)
print(events.columns.tolist())

print("\nSentiment coverage:")
print("with sentiment:", events["sentiment_score"].notna().sum())
print("missing sentiment:", events["sentiment_score"].isna().sum())

print("\nRows by ticker:")
print(events["ticker"].value_counts().sort_index())

print("\nLLM model:")
print(events["llm_model"].value_counts(dropna=False))

print("\nArticle types:")
print(events["article_type"].value_counts(dropna=False).head(20))

print("\nSentiment score:")
print(events["sentiment_score"].describe())

display(events.sample(20, random_state=42))
```
